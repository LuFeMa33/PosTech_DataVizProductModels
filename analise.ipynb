```python
df_obesidade = pd.read_csv('https://raw.githubusercontent.com/LuFeMa33/PosTech_DataVizProductModels/refs/heads/main/Obesity.csv')
```

```python
df_obesidade.shape
```

```python
df_obesidade.info()
```

```python
dict_colunas_alteracao = {'Gender':'Genero',
                          'Age':'Idade',
                          'Height':'Altura',
                          'Weight':'Peso',
                          'family_history':'historico_familiar',
                          'FAVC':'frequencia_consumo_alta_caloria',
                          'FCVC':'frequencia_consumo_vegetais',
                          'NCP':'numero_refeicoes_principais',
                          'CAEC':'consumo_lanches_entre_refeicoes',
                          'SMOKE':'fumante',
                          'CH2O':'consumo_agua',
                          'SCC':'monitoramento_consumo_calorias',
                          'FAF':'frequencia_atividade_fisica',
                          'TUE':'tempo_uso_dispositivos_eletronicos',
                          'CALC':'consumo_alcool',
                          'MTRANS':'meio_transporte',
                          'Obesity':'Obesidade'}
```

```python
df_obesidade.rename(columns=dict_colunas_alteracao, inplace=True)
```

```python
df_obesidade.head()
```

```python
df_obesidade['historico_familiar'] = df_obesidade['historico_familiar'].map({'yes': '1', 'no': '0'})
df_obesidade['frequencia_consumo_alta_caloria'] = df_obesidade['frequencia_consumo_alta_caloria'].map({'yes': '1', 'no': '0'})
df_obesidade['fumante'] = df_obesidade['fumante'].map({'yes': '1', 'no': '0'})
df_obesidade['monitoramento_consumo_calorias'] = df_obesidade['monitoramento_consumo_calorias'].map({'yes': '1', 'no': '0'})
df_obesidade['Indice_Genero'] = df_obesidade['Genero'].map({'Female':0, 'Male':1})
```
