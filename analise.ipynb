```python
dict_colunas_alteracao = {'Gender':'Genero',
                          'Age':'Idade',
                          'Height':'Altura',
                          'Weight':'Peso',
                          'family_history':'historico_familiar',
                          'FAVC':'frequencia_consumo_alta_caloria',
                          'FCVC':'frequencia_consumo_vegetais',
                          'NCP':'numero_refeicoes_principais',
                          'CAEC':'consumo_lanches_entre_refeicoes',
                          'SMOKE':'fumante',
                          'CH2O':'consumo_agua',
                          'SCC':'monitoramento_consumo_calorias',
                          'FAF':'frequencia_atividade_fisica',
                          'TUE':'tempo_uso_dispositivos_eletronicos',
                          'CALC':'consumo_alcool',
                          'MTRANS':'meio_transporte',
                          'Obesity':'Obesidade'}
```

```python
df_obesidade.rename(columns=dict_colunas_alteracao, inplace=True)
```

```python
df_obesidade['historico_familiar'] = df_obesidade['historico_familiar'].map({'yes': '1', 'no': '0'})
df_obesidade['frequencia_consumo_alta_caloria'] = df_obesidade['frequencia_consumo_alta_caloria'].map({'yes': '1', 'no': '0'})
df_obesidade['fumante'] = df_obesidade['fumante'].map({'yes': '1', 'no': '0'})
df_obesidade['monitoramento_consumo_calorias'] = df_obesidade['monitoramento_consumo_calorias'].map({'yes': '1', 'no': '0'})
df_obesidade['Indice_Genero'] = df_obesidade['Genero'].map({'Female':0, 'Male':1})
```

```python
X = df_obesidade_sem_imc[['Idade',
 'Altura',
 'Peso',
 'historico_familiar',
 'frequencia_consumo_alta_caloria',
 'frequencia_consumo_vegetais',
 'numero_refeicoes_principais',
 'fumante',
 'consumo_agua',
 'monitoramento_consumo_calorias',
 'frequencia_atividade_fisica',
 'tempo_uso_dispositivos_eletronicos',
 'Indice_Genero',
 'consumo_lanches_entre_refeicoes_Frequently',
 'consumo_lanches_entre_refeicoes_Sometimes',
 'consumo_lanches_entre_refeicoes_no',
 'consumo_alcool_Frequently',
 'consumo_alcool_Sometimes',
 'consumo_alcool_no',
 'meio_transporte_Bike',
 'meio_transporte_Motorbike',
 'meio_transporte_Public_Transportation',
 'meio_transporte_Walking']]
```

```python
y = df_obesidade_sem_imc['Obesidade']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state = 101)
```

```python
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
fig = plt.figure(figsize=(8, 5), dpi=80)
plt.rcParams.update({'font.size': 14})
ax = sns.regplot(x = y_test, y = y_pred)
ax.set(xlabel='B real', ylabel='B predito')
ax = plt.plot(y_test, y_test, '--r')
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
MAE = mean_absolute_error(y_test, y_pred)
MSE = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print('MAE', MAE)
print('MSE', MSE)
print('r²', r2)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model_dtr = DecisionTreeRegressor(random_state=101, max_depth=10)
model_dtr.fit(X_train, y_train)
```

```python
y_pred_model_dtr = model_dtr.predict(X_test)
```

```python
MAE = mean_absolute_error(y_test, y_pred_model_dtr)
MSE = mean_squared_error(y_test, y_pred_model_dtr)
r2 = r2_score(y_test, y_pred_model_dtr)

print('MAE', MAE)
print('MSE', MSE)
print('r²', r2)
```

```python
fig = plt.figure(figsize=(8, 5), dpi=80)
plt.rcParams.update({'font.size': 14})
ax = sns.regplot(x = y_test, y = y_pred_model_dtr)
ax.set(xlabel='y real', ylabel='y predito')
ax = plt.plot(y_test, y_test, '--r')
plt.show()
```
